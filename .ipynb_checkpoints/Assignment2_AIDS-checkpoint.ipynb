## Question 1: Student Result Checker 
A school wants to automatically determine whether a student has passed or failed. 
Ask the user to enter the student's marks. 
- If marks are 50 or above, display "Pass".

## Question 2: Electricity Bill Category 
An electricity company wants to categorize customers based on their monthly electricity units. 
Ask the user to enter the number of units consumed. 
- Less than 100 units → "Low Usage"  
- 100 to 300 units → "Medium Usage"  
- More than 300 units → "High Usage"  
Display the appropriate category. 

```python
unitsConsumed = int(input("Enter number of units consumed:"))
if unitsConsumed < 100:
    category = "Low Usage"
elif unitsConsumed >= 100 and unitsConsumed <= 300:
    category = "Medium Usage"
else:
    category = "High Usage"
print(f"Since you consumed {unitsConsumed} units you are in {category} category.")
```

## Question 3: Online Shopping Discount 
An online store gives discounts based on the total purchase amount. 
Ask the customer for the total amount. 
- Less than 5,000 → No discount  
- 5,000 to 9,999 → 10% discount  
- 10,000 to 19,999 → 15% discount  
- 20,000 or above → 20% discount  
Calculate and display: 
- Original amount  
- Discount  
- Final amount  

```python
totalAmount = int(input("Enter total amount of bill:"))
if totalAmount < 5000:
    discount = 0
    finalAmount = totalAmount
elif totalAmount < 10000:
    discount = 10
    finalAmount = totalAmount - (totalAmount * 0.1)
elif totalAmount < 20000:
    discount = 15
    finalAmount = totalAmount - (totalAmount * 0.15)
else:
    discount = 20
    finalAmount = totalAmount - (totalAmount * 0.2)
print(f"You have purchased items of total amount {totalAmount} so you get discount {discount}%. You're total bill after discount of Rs {totalAmount * (discount / 100)} is Rs{finalAmount}.")
```

## Question 4: Employee Salary Grade 
A company wants to assign salary grades based on an employee's salary. 
Ask the employee to enter their salary. 
- Below 30,000 → Grade C  
- 30,000 to 59,999 → Grade B  
- 60,000 to 99,999 → Grade A  
- 100,000 or above → Grade A+  
Display the employee's salary grade. 

```python
salary = int(input("Enter your salary:"))
if salary < 30000:
    grade = "C"
elif salary < 60000:
    grade = "B"
elif salary < 100000:
    grade = "A"
else:
    grade = "A+"

print(f"Your salary is {salary} so you're grade is {grade}")
```

## Question 7: Number Analyzer 
A data-entry application needs to analyze a number entered by the user. 
Ask the user to enter a number. 
Your program should determine: 
- Whether the number is positive, negative, or zero.  
- If it is positive or negative, also determine whether it is even or odd.  
For example: 
Enter number: -7 
Negative 
Odd 

```python
num = int(input("Enter a number:"))
if num < 0:
    print("Negative")
    if num % 2 == 0:
        print("Even")
    else:
        print("Odd")
elif num > 0:
    print("Positive")
    if num % 2 == 0:
        print("Even")
    else:
        print("Odd")
else:
    print("Number is Zero")
```

## Question 8: ATM Withdrawal 
You are developing an ATM withdrawal system. 
The user's account balance is: 
50000 
Ask the user to enter the withdrawal amount. 
Apply these rules: 
- If the amount is less than or equal to 0, display "Invalid Amount".  
- If the amount is greater than the balance, display "Insufficient Balance".  
- Otherwise, process the withdrawal and display the remaining balance. 

```python
balance = 50000
amount = int(input("Enter the amount for withdrawal:"))
if amount <= 0:
    print("Invalid Amount")
elif amount > balance:
    print("Insufficient Balance")
else:
    balance -= amount
    print(f"Withdrawal success of Rs{amount}. You have remaining balance of Rs{balance}")
```

## Question 9: University Admission Eligibility

## Question 10: Job Eligibility System 
A company wants to filter candidates for a job. 
Ask the candidate for: 
- Age  
- Years of experience  
- Technical test score  
Eligibility rules: 
- Age must be at least 18.  
- Experience must be at least 1 year.  
- Technical test score must be at least 60.  
If all requirements are satisfied, display: 
Candidate is Eligible 
Otherwise, display the specific reason why the candidate is not eligible

```python
age =  int(input("Enter your age:"))
experience = float(input("Enter how many years of experience do you have:"))
testScore = int(input("Enter the score of technical test:"))
if age >= 18 and experience >= 1 and testScore >= 60:
    print("Congrats you're eligible for job.")
elif age < 18:
    print("You're not eligible for this job because you are underage.")
elif experience < 1: 
    print("You're not eligible for this job because your experience is too low.")
else:
    print("You're not eligible for this job because your technical test score is lower than 60.")
```

## Question 11: Smart Student Admission System 
A university wants to create a student admission screening system. 
Ask the student to enter: 
- Student name  
- Age  
- Matric percentage  
- Intermediate percentage  
- Entry test score  
- Whether the student has a domicile (yes/no)  
The university has the following admission rules: 
Basic Eligibility 
The student must: 
- Be at least 17 years old.
-  Have at least 60% in Matric.  
-  Have at least 60% in Intermediate.  
- Have at least 50% in the entry test.  
- Have a valid domicile.  
If the student fails any requirement, display Not Eligible and clearly mention the reason. 

```python
studentName = input("Enter student name: ")
age = int(input("Enter age of the student: "))
matricPercentage = float(input("Enter Matric percentage (%): "))
interPercentage = float(input("Enter Intermediate percentage (%): "))
entryTestScore = float(input("Enter entry test score (%): "))
domicile = input("Do you have a domicile (yes/no): ").lower().strip()

status = "Not Eligible"
reason = ""
merit = "None"
scholarship = "No Scholarship"

if age < 17:
    reason = "Age must be at least 17."
elif matricPercentage < 60:
    reason = "Matric percentage must be at least 60%."
elif interPercentage < 60:
    reason = "Intermediate percentage must be at least 60%."
elif entryTestScore < 50:
    reason = "Entry test score must be at least 50%."
elif domicile != "yes":
    reason = "A valid domicile is required."
else:
    status = "Eligible"

    if entryTestScore >= 85 and interPercentage >= 85:
        merit = "Excellent"
        scholarship = "50%"
    elif entryTestScore >= 70 and interPercentage >= 70:
        merit = "Good"
        scholarship = "25%"
    else:
        merit = "Normal"

print("\n===== ADMISSION REPORT =====")
print(f"""Name: {studentName}
Age: {age}
Intermediate: {interPercentage}%
Entry Test: {entryTestScore}%
Status: {status}""")
if status == "Not Eligible":
    print(f"Reason: {reason}")
print(f"Merit: {merit} Merit")
print(f"Scholarship: {scholarship}")
```

## Question 12: Employee Performance & Promotion System 
A company wants to automatically evaluate its employees for promotion. 
Ask the employee for: 
- Employee name  
- Years of experience  
- Monthly salary  
- Performance score out of 100  
- Number of projects completed  
- Number of late days  
Promotion Rules 
An employee can be promoted only if: 
- Experience is at least 2 years.  
- Performance score is at least 70.  
- At least 3 projects have been completed.  
- Late days are less than 10.  
If any requirement is not satisfied, the employee is Not Eligible for Promotion. 
The program should display the specific reason. 

```python
employeeName = input("Enter the name of employee: ")
yearsOfExperience = int(input("How many years of experience do you have? "))
monthlySalary = int(input("What is your monthly salary? "))
performanceScore = int(input("What is your performance score out of 100? "))
projectsCompleted = int(input("How many projects have you completed? "))
lateDays = int(input("How many days have you been late? "))
promotionStatus = "Not Eligible"
promotionLevel = "None"
bonus = 0
reason = ""

if yearsOfExperience < 2:
    reason = "Experience must be at least 2 years."
elif performanceScore < 70:
    reason = "Performance score must be at least 70."
elif projectsCompleted < 3:
    reason = "At least 3 projects must be completed."
elif lateDays >= 10:
    reason = "Late days must be less than 10."
else:
    promotionStatus = "Eligible"
    if performanceScore >= 90 and yearsOfExperience >= 5:
        promotionLevel = "Senior"
    elif performanceScore >= 80 and yearsOfExperience >= 3:
        promotionLevel = "Mid-Level"
    else:
        promotionLevel = "Junior"
    if performanceScore >= 90:
        bonus = 20
    elif performanceScore >= 80:
        bonus = 15
    else:
        bonus = 10
print("\n===== EMPLOYEE REPORT =====")
print(f"Employee: {employeeName}")
print(f"Experience: {yearsOfExperience} years")
print(f"Performance: {performanceScore}")
print(f"Projects: {projectsCompleted}")
print(f"Late Days: {lateDays}")
print(f"Promotion Status: {promotionStatus}")

if promotionStatus == "Not Eligible":
    print(f"Reason: {reason}")
else:
    print(f"Promotion Level: {promotionLevel}")
    print(f"Bonus: {bonus}%")
```
