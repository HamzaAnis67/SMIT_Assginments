### Question 3. Arithmetic Operations 
- Create two number variables and calculate their:
  
• Addition 
• Subtraction 
• Multiplication 
• Division 
• Modulus 
• Power

```python
num1 = int(input("Enter first number:"))
num2 = int(input("Enter second number:"))
addition = num1 + num2
difference = num1 - num2
product = num1 * num2
division = num1 / num2
mod = num1 % num2
power = num1 ** num2

print(f"The Result of additon of two number is: {addition}")
print(f"The Result of subtraction of two number is: {difference}")
print(f"The Result of multiplication of two number is: {product}")
print(f"The Result of division of two number is: {division}")
print(f"The Result of modulus of two number is: {mod}")
print(f"The Result of power of two number is: {power}")
```

### Question 4. Student Marks Calculator 
- Store the marks of three subjects in separate variables. Calculate and print the total 
marks and average marks. 

```python
dsa = float(input("Enter marks obtained in Data Structures and Algorithm:"))
coal = float(input("Enter marks obtained in Computer Organization and Assembly Language:"))
se = float(input("Enter marks obtained in Software Engineering:"))
totalMarks = dsa + coal + se
avgMarks = totalMarks / 3
print(f"Total Marks obtained of all three subjects are: {totalMarks} and Average Marks  is: {avgMarks}.")
```

### Question 6. Rectangle Area 
- Create variables for the length and width of a rectangle. Calculate and print its area and 
perimeter. 

```python
length = float(input("Enter length of the rectangle:")) 
width = float(input("Enter width of the rectangle:"))
area = length * width
perimeter = 2 * (length + width)
print(f"The area of rectangle is: {area} and perimeter of rectangle is: {perimeter}")
```

### Question 7. Time Converter 
- Store 250 minutes in a variable. Calculate and print the number of hours and remaining 
minutes. 

```python
minutes = 250
hours = minutes // 60
remainingMinutes = minutes % 60
print(f"There are {hours} hours and {remainingMinutes} minutes in {minutes} minutes.")
```

### Question 8. Student Eligibility 
- Create variables for marks and attendance. 
A student is eligible if:

• Marks are greater than or equal to 50  
• Attendance is greater than or equal to 75  
Use logical operators to calculate and print whether the student is eligible. 

```python
marks = float(input("Enter your marks:"))
attendance = float(input("Enter attendance in percentage (%):"))
isEligible = (marks >= 50) and (attendance >= 75) 
print(f"The following student is eligible: {isEligible}")
```
