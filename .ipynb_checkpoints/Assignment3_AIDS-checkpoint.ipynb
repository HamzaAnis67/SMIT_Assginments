## Question 3: Student Marks Analyzer 
A teacher wants to enter marks for 5 students. 
Use a loop to: 
1. Ask for each student's marks.  
2. Display whether the student passed or failed.  
3. Count how many students passed.  
4. Count how many students failed.  
At the end, display: 
Total Passed: X 
Total Failed: Y

```python
totalPassed = 0
totalFailed = 0
for i in range(1,6):
    studentMarks = int(input(f"Enter marks of Student {i}:"))
    if studentMarks >= 50:
        print("Student Passed")
        totalPassed += 1
    else:
        print("Student Failed")
        totalFailed += 1
print(f"Total Passed: {totalPassed}")
print(f"Total Failed: {totalFailed}")
```

## Question 5: Number Analyzer 
A user wants to enter 10 numbers. 
Use a for loop to analyze each number. 
Your program should count: 

• How many numbers are positive  
• How many numbers are negative  
• How many numbers are zero  
At the end, display all three counts. 
Example: 
Positive Numbers: 4 
Negative Numbers: 5 
Zeroes: 1

```python
positiveNumber = 0
negativeNumber = 0
zero = 0
for i in range(10):
    num = int(input("Enter a number:"))
    if num == 0:
        zero +=1
    elif num < 0:
        negativeNumber += 1
    else:
        positiveNumber += 1

print(f"""\nPositive Numbers: {positiveNumber}
Negative Numbers: {negativeNumber}
Zeroes: {zero}""")
```

## Question 7: ATM Menu 
Create a simple ATM system using a while loop. 
The initial balance is: 
50000 
Display this menu repeatedly: 
1. Check Balance 
2. Deposit 
3. Withdraw 
4. Exit 

The program should continue until the user selects 4. 
Implement the following: 
- Check balance  
- Deposit money  
- Withdraw money  
- Prevent withdrawal if the amount is greater than the balance  
- Exit the program  

```python
balance = 50000
while True:
    option = int(input("Enter 1 to check balance, 2 to deposit money, 3 to withdraw money or 4 to exit"))
    if option == 1:
        print(f"Your current balance is {balance}")
    elif option == 2:
        depositAmount = int(input("Enter how much money do you want to deposit?"))
        if depositAmount <= 0:
            print("Invalid amount entered")
        else:
            balance += depositAmount
            print(f"Deposit successful your current balance is {balance}.")
    elif option == 3:
        withdrawAmount = int(input("Enter how much money do you want to withdraw?"))
        if withdrawAmount > balance or withdrawAmount <= 0:
            print("Insufficient balance or Invalid Amount")
        else:
            balance -= withdrawAmount
            print(f"Withdrawl successful your current balance is {balance}.")
    elif option == 4:
        break
    else:
        print("Invalid number entered. Please enter number between 1-4")
```

## Question 8: Number Guessing Game 
Create a number guessing game. 
Set the secret number to: 
secret_number = 25 
Ask the user to guess the number. 
Use a while loop. 
If the guess is: 
- Greater than 25 → "Too High"  
- Less than 25 → "Too Low"  
- Equal to 25 → "Correct!"  
Also count how many attempts the user made. 
Example: 
Guess: 30 
Too High 
Guess: 20 
Too Low 
Guess: 25 
Correct! 
Attempts: 3

```python
secret_number = 25 
attempts = 0
while True:
    attempts += 1
    guess = int(input("Guess the secret number:"))
    if guess == secret_number:
        print(f"Correct! \nAttempts: {attempts}")
        break
    elif guess > secret_number:
        print("Too High")
    else:
        print("Too Low")
```

## Question 9: Student Marks Entry System 
A teacher wants to enter student marks continuously. 
Use a while loop. 
The program should keep asking for marks until the teacher enters: -1 
For every student's marks: 
- 50 or above → Pass  
- Below 50 → Fail  
At the end, display: 
Total Students 
Total Passed 
Total Failed 
The -1 should not be counted as a student's marks.

```python
totalStudents = 0 
totalPassed = 0
totalFailed = 0
while True:
    studentMarks = int(input("Enter student marks or -1 to exit"))
    if studentMarks == -1:
        break
    else:
        totalStudents += 1
        if studentMarks >= 50:
            totalPassed += 1
        else:
            totalFailed += 1
print("\n====Students Record====")            
print(f"Total Students: {totalStudents}")
print(f"Total Passed: {totalPassed}")
print(f"Total Failed: {totalFailed}")
```

## Question 10: Restaurant Ordering System 
You are developing a simple restaurant ordering system. 
Display this menu repeatedly: 
1. Burger - Rs. 500 
2. Pizza - Rs. 1000 
3. Fries - Rs. 300 
4. Drink - Rs. 150 
5. Checkout 
Use a while loop. 
When the customer selects an item: 
- Add its price to the total bill.  
- Display the selected item.  
- Continue showing the menu.  
When the customer selects 5, stop taking orders and display the final bill. 
Example: 
Burger added 
Pizza added 
Drink added 
Total Bill = Rs. 1650

```python
totalBill = 0
while True:
    item = int(input("""Choose any item from below or select 5 to checkout
    1. Burger - Rs. 500 
    2. Pizza - Rs. 1000 
    3. Fries - Rs. 300 
    4. Drink - Rs. 150 
    5. Checkout
    """))
    if item == 1:
        totalBill += 500
        print("Burger added")
    elif item == 2:
        totalBill += 1000
        print("Pizza added")
    elif item == 3:
        totalBill += 300
        print("Fries added")
    elif item == 4:
        totalBill += 150
        print("Drink added")
    elif item == 5:
        break
    else:
        print("Invalid number entered.")
print(f"Total Bill = Rs.{totalBill}")
```

## Question 11: Student Performance Management System 
A teacher wants to create a system for analyzing the performance of multiple students. 
First ask: 
Enter number of students: 
For each student, ask: 
- Student name  
- Marks in Python  
- Marks in Mathematics  
- Marks in English  
Calculate: 
Total Marks 
Percentage 
Result Rules 
A student is considered Pass only if: 
- Overall percentage is at least 50%  
- AND marks in every subject are at least 40.  
Otherwise, the student is Fail.

```python
totalPassed = 0
totalFailed = 0
highestPercentage = 0
lowestPercentage = 100
classTotalMarks = 0
classTotalPercentage = 0
totalStudents = int(input("Enter number of students:"))
if totalStudents <= 0:
    print("Please enter a number of students greater than 0.")
else:
    for i in range(totalStudents):
        studentName = input("Enter student name:")  
        marksPython = int(input("Enter marks obtained in Python"))
        marksMathematics = int(input("Enter marks obtained in Mathematics"))
        marksEnglish = int(input("Enter marks obtained in English"))
        totalMarks = marksEnglish + marksMathematics + marksPython
        percentage = 100 * (totalMarks / 300)
        classTotalMarks += totalMarks
        classTotalPercentage += percentage    
        if i == 0:
            highestPercentage = percentage
            lowestPercentage = percentage
        else:
            if percentage > highestPercentage:
                    highestPercentage = percentage
            if percentage < lowestPercentage:
                    lowestPercentage = percentage
        if percentage >= 50 and marksEnglish >= 40 and marksMathematics >= 40 and marksPython >= 40:
            totalPassed += 1
            if percentage >= 90:
                Grade = "A+"
            elif percentage >= 80:
                Grade = "A"
            elif percentage >= 70:
                Grade = "B"
            elif percentage >= 60:
                Grade = "C"
            elif percentage >= 50:
                Grade = "D"
        else:
            totalFailed += 1
print("\n===== CLASS REPORT =====")
print(f"""Total Students: {totalStudents} 
Passed: {totalPassed}
Failed: {totalFailed}
Highest Percentage: {highestPercentage}% 
Lowest Percentage: {lowestPercentage}% 
Class Average: {(classTotalPercentage/totalStudents)} """)
```

## Question 12: Restaurant Ordering & Billing System 
You are developing a restaurant ordering system. 
The restaurant has the following menu: 
MENU  
1. Burger    Rs. 500 
2. Pizza    Rs. 1000  
3. Biryani  Rs. 350      
4. Fries    Rs. 250    
5. Drink    Rs. 150     
6. Checkout 

The program should repeatedly display the menu and allow the customer to order multiple 
items. </br>
The program should continue until the customer selects: 

6. Checkout </br>
When an item is selected</br>
Add its price to the total bill. </br>
Also keep track of how many items the customer ordered. </br>
For example: </br>
Burger added </br>
Pizza added </br>
Drink added </br>
Invalid Input </br>
If the customer enters something other than 1-6: </br>
Invalid choice. Please try again. </br>
The program should not terminate. </br>

```python
totalBill = 0
totalItems = 0
while True:
    item = int(input("""MENU  
    1. Burger    Rs.500
    2. Pizza    Rs. 1000  
    3. Biryani  Rs. 350      
    4. Fries    Rs. 250    
    5. Drink    Rs. 150        
    6. Checkout 
    Enter what item do you want to order?"""))
    if item == 1:
        totalBill += 500
        totalItems += 1 
        print("Burger added")
    elif item == 2:
        totalBill += 1000
        totalItems += 1
        print("Pizza added")
    elif item == 3:
        totalBill += 350
        totalItems += 1
        print("Briyani added")
    elif item == 4:
        totalBill += 250
        totalItems += 1
        print("Fries added")
    elif item == 5:
        totalBill += 150
        totalItems += 1
        print("Drink added")
    elif item == 6:
        break
    else:
        print("Invalid choice. Please try again. ")
if totalBill >= 5000:
    discount = (totalBill*0.2)
    finalBill = totalBill- discount
elif totalBill >= 3000:
    discount = (totalBill*0.15)
    finalBill = totalBill- discount
elif totalBill >= 1500:
    discount = (totalBill*0.1)
    finalBill = totalBill- discount
else: 
    discount = 0
    finalBill = totalBill

if finalBill >= 3000:
    deliveryCharges = 0
else: 
    deliveryCharges = 200
    finalBill += deliveryCharges

print("\n===== FINAL BILL =====")
print(f"""Total Items: {totalItems}
Original Bill: Rs. {totalBill} 
Discount: Rs. {discount} 
Delivery Charges: Rs. {deliveryCharges} 
Final Bill: Rs. {finalBill} 
Thank you for ordering! """)
    
```
